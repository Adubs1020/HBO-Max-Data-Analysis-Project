### Selected all the TV Shows from the hbo_content table 

```python
%%sql SELECT *
FROM hbo_content
WHERE type = 'TV'
ORDER BY index
```

### Selected all the TV Shows from the hbo_max_content table

```python
%%sql SELECT *
FROM hbo_max_content
WHERE type = 'TV'
ORDER BY index;
```

###  Compare the average imdb score for ratings of HBO TV Shows that were released in the 2000s

```python
%%sql SELECT rating, AVG(imdb_score) as avg_imdb_score
FROM hbo_content
WHERE type = 'TV'
AND decade LIKE '2%'
AND rating != 'NA'
GROUP BY rating
ORDER BY avg_imdb_score DESC;
```

###  Compare the average rotten score for ratings of HBO TV Shows that were released in the 2000s

```python
%%sql SELECT rating, AVG(rotten_score) as avg_rotten_score
FROM hbo_content
WHERE type = 'TV'
AND decade LIKE '2%'
AND rating != 'NA'
GROUP BY rating
ORDER BY avg_rotten_score DESC;
```

###  List the top 10 HBO MA TV shows by imdb_score that were released in the 2000s. 

```python
%%sql SELECT DISTINCT title,type,year, imdb_score
FROM hbo_content
WHERE type = 'TV'
AND decade LIKE '2%'
AND imdb_score IS NOT NULL
ORDER BY imdb_score DESC
LIMIT 10;
```

###  List the top 10  HBO MA TV shows by rotten_score that were released in the 2000s. 


```python
%%sql SELECT DISTINCT title, type,year, rotten_score
FROM hbo_content
WHERE type = 'TV'
AND decade LIKE '2%'
AND rotten_score IS NOT NULL
ORDER BY rotten_score DESC
LIMIT 10;
```

### List the HBO dramas that have a imdb score higher than the average  imdb score  and a rotten score higher     than the average rotten score 

```python
%%sql SELECT title, ROUND(AVG(imdb_score), 2) as avg_imdb_score, ROUND(AVG(rotten_score), 2) as avg_rotten_score
FROM hbo_content
WHERE type = 'TV'
AND genres_drama = True
GROUP BY title
HAVING AVG(imdb_score) > (SELECT AVG(imdb_score) FROM hbo_content)
AND AVG(rotten_score) > (SELECT AVG(rotten_score) FROM hbo_content)
ORDER BY avg_imdb_score DESC, avg_rotten_score DESC
```

###  List the HBO comedies that have a imdb score higher than the average  imdb score and a rotten score higher than the average rotten score 

```python
%%sql SELECT title, ROUND(AVG(imdb_score), 2) as avg_imdb_score, ROUND(AVG(rotten_score), 2) as avg_rotten_score
FROM hbo_content
WHERE type = 'TV'
AND genres_comedy = True
GROUP BY title
HAVING AVG(imdb_score) > (SELECT AVG(imdb_score) FROM hbo_content)
AND AVG(rotten_score) > (SELECT AVG(rotten_score) FROM hbo_content)
ORDER BY avg_imdb_score DESC, avg_rotten_score DESC
```

###  List the shows that are on HBO Max but not on HBO that have an imdb score of 9.0 or higher

```python
%%sql SELECT DISTINCT title, year, rating, imdb_score
FROM hbo_max_content
WHERE type = 'TV'
AND imdb_score >= 8.00
AND title NOT IN (SELECT title
			      FROM hbo_content)
ORDER BY imdb_score DESC;
```

### List the shows that are on HBO Max but not on HBO that have a rotten score of 80.0 or higher


```python
%%sql SELECT DISTINCT title, year, rating, rotten_score
FROM hbo_max_content
WHERE type = 'TV'
AND rotten_score >= 80.0
AND title NOT IN (SELECT title
				  FROM hbo_content)
ORDER BY rotten_score DESC;
```
